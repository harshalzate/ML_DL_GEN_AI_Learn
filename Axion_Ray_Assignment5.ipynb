## Clean Task 2 Dataset

### Subtask:
Perform data cleaning on the Task 2 dataset (df_task2), including handling missing values, addressing duplicate entries, ensuring consistent data types, and standardizing text fields.


### Reasoning for Cleaning df_task2:

The previous analysis identified `df_task2` as a new dataset requiring cleaning to integrate with the existing `df` and prepare for further analysis. The cleaning process will focus on:

1.  **Handling Duplicates**: Removing duplicate rows is crucial to ensure each record is unique and does not skew subsequent analyses.
2.  **Imputing Missing Values**: The 'Cause' and 'Correction' columns have a significant number of missing values. Imputing these with 'unknown' is appropriate given their text-based categorical nature, ensuring no loss of information while allowing these records to be included in analyses.
3.  **Standardizing Text Fields**: Converting 'object' type columns to lowercase and stripping whitespace will ensure consistency across categorical entries, preventing issues with unique value counts and matching during merges.
4.  **Correcting 'Model Year'**: The 'Model Year' column, if containing '0' values, indicates incorrect or placeholder data. Replacing these with the mode of the column will provide a more realistic and statistically sound imputation for a numerical/categorical field.
5.  **Verification**: After applying cleaning steps, it's essential to verify their effectiveness by checking for remaining missing values and reviewing data types to confirm data integrity.

**Reasoning**:
Now that the reasoning for cleaning `df_task2` is established, I will implement the cleaning steps outlined in the instructions: removing duplicates, handling missing values in 'Cause' and 'Correction' columns, standardizing text fields, and correcting 'Model Year' entries, followed by verification.



````python
print("Starting data cleaning process for df_task2...")

# 1. Check for and remove duplicate rows
initial_rows = len(df_task2)
df_task2.drop_duplicates(inplace=True)
duplicates_removed = initial_rows - len(df_task2)
if duplicates_removed > 0:
    print(f"Removed {duplicates_removed} duplicate rows from df_task2.")
else:
    print("No duplicate rows found in df_task2.")

# 2. Handle missing values in 'Cause' and 'Correction' columns with 'unknown'
for col in ['Cause', 'Correction']:
    if col in df_task2.columns:
        missing_count = df_task2[col].isnull().sum()
        if missing_count > 0:
            df_task2[col] = df_task2[col].fillna('unknown')
            print(f"Imputed {missing_count} missing values in '{col}' with 'unknown'.")
        else:
            print(f"No missing values in '{col}'.")

# 3. Standardize text data in all 'object' type columns
object_cols_task2 = df_task2.select_dtypes(include='object').columns
for col in object_cols_task2:
    df_task2[col] = df_task2[col].astype(str).str.strip().str.lower()
print("Standardized text in all object type columns in df_task2 (stripped whitespace and converted to lowercase).")

# 4. Correct 'Model Year' entries where value is 0 by replacing with the mode
if 'Model Year' in df_task2.columns and (df_task2['Model Year'] == 0).any():
    mode_model_year = df_task2[df_task2['Model Year'] != 0]['Model Year'].mode()[0]
    df_task2['Model Year'] = df_task2['Model Year'].replace(0, mode_model_year)
    print(f"Replaced '0' values in 'Model Year' with its mode ({mode_model_year}).")
else:
    print("No '0' values to correct or 'Model Year' column not found.")

# 5. Verify no remaining missing values and check data types
missing_values_after_cleaning_task2 = df_task2.isnull().sum().sum()
if missing_values_after_cleaning_task2 == 0:
    print("Verification successful: No remaining missing values in df_task2.")
else:
    print(f"Verification failed: {missing_values_after_cleaning_task2} missing values still present in df_task2. Details:\n{df_task2.isnull().sum()[df_task2.isnull().sum() > 0]}")

print("Data cleaning process for df_task2 completed.")

# Display info and head after cleaning for a quick check
print("\nInformation about df_task2 after cleaning:")
df_task2.info()
print("\nFirst 5 rows of df_task2 after cleaning:")
display(df_task2.head())
````
